#### 🔀 Bitwise OR Operator (`|`)

The **Bitwise OR (`|`) operator** compares the corresponding bits of two integers. It returns **`1` if at least one of the corresponding bits is `1`**. It returns `0` only when **both bits are `0`**.

The Bitwise OR operator is commonly used for **setting bits**, **enabling flags**, **permission management**, **network programming**, and **embedded systems**.

#### 📖 Topics Covered

##### 1. Introduction to the Bitwise OR Operator

The Bitwise OR operator compares every corresponding bit of two integers.

A bit is set to **1 if either or both bits are 1**.

Example:

```text
12 = 1100
10 = 1010
--------------
     1110
```

Result:

```text
14
```

##### 10. Common Mistakes

Avoid these common mistakes while using the Bitwise OR operator.

##### Confusing `|` with `or`

Incorrect for logical conditions:

```python
True | False
```

Preferred:

```python
True or False
```

> **Why?** `|` performs a bitwise operation, while `or` performs a logical operation.

---

##### Using Floating-Point Numbers

Incorrect

```python
10.5 | 4
```

##### Output

```text
TypeError
```

The `|` operator works only with integers.

---

##### Forgetting Binary Representation

Incorrect assumption:

```python
12 | 10
```

The operation is performed on the binary representation of the numbers, not on their decimal values.

---

##### Using Strings

Incorrect

```python
"12" | "10"
```

##### Output

```text
TypeError
```

##### 11. Best Practices

Follow these recommended practices when using the Bitwise OR operator.

- Understand binary representation before using bitwise operators.
- Use `bin()` to visualize bit patterns.
- Use meaningful variable names such as `flags`, `mask`, and `permissions`.
- Add comments when manipulating bits.
- Use `or` for Boolean logic and `|` for bitwise operations.
- Keep expressions simple and readable.

````python
# Basic Bitwise OR
result = 10 | 6

print(result)
print('''
10 = 1010
 6 = 0110
-----------
14 = 1110
''')
````

````python
# Bitwise OR
a = 12
b = 10

result = a | b
print(result)

print('''
12 = 1100
10 = 1010
-----------
14 = 1110
''')
````

````python
# Bitwise OR
a = 15
b = 7

result = a | b
print(result)
print('''
15 = 1111
 7 = 0111
-----------
15 = 1111
''')
````

````python
# Binary Representation
a = 10
b = 6

result = a | b

print(bin(a))
print(bin(b))
print(bin(result))
````

````python
# OR with Zero
number = 25

result = number | 0
print(result)
````

````python
# OR with Itself
number = 25

result = number | number
print(result)
````

````python
# OR with One
number = 10

result = number | 1
print(result)

print('''
10 = 1010
 1 = 0001
-----------
11 = 1011
''')
````

````python
# Binary Literals
a = 0b1100
b = 0b1010

result = a | b

print(bin(result))
print(result)
````

````python
# Variables
a = 20
b = 12

result = a | b
print(result)

print('''
20 = 10100
12 = 01100
------------
28 = 11100
''')
````

````python
# Multiple OR Operations

result = 8 | 4 | 2 | 1
print(result)

print('''
8 | 4 = 12
12 | 2 = 14
14 | 1 = 15
''')
````

````python
# Variable and Mask
number = 8
mask = 7

result = number | mask
print(result)

print('''
 8 = 1000
 7 = 0111
-----------
15 = 1111
''')
````

````python
# Bit Mask
number = 8
mask = 3

result = number | mask
print(result)

print('''
 8 = 1000
 3 = 0011
-----------
11 = 1011
''')
````

````python
# Set Bit Position
number = 8
position = 2

mask = 1 << position
result = number | mask

print(result)
print('''
1000
0100
----
1100
''')
````

````python
# Set Existing Bit
number = 12
position = 2

mask = 1 << position
result = number | mask

print(result)
print('''
1100
0100
----
1100
''')
````

````python
# Set Multiple Bits
number = 8
mask = 7

result = number | mask
print(result)

print('''
 8 = 1000
 7 = 0111
-----------
15 = 1111
''')
````

````python
# Permissions

READ = 1
WRITE = 2
EXECUTE = 4

permissions = READ | WRITE
print(permissions)

print('''
READ  = 001
WRITE = 010
--------------
        011
''')
````

````python
# Add Execute Permission

READ = 1
WRITE = 2
EXECUTE = 4

permissions = READ | WRITE
permissions = permissions | EXECUTE

print(permissions)
print('''
011
100
---
111
''')
````
