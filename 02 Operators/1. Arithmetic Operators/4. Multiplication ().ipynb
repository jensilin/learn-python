#### ✖️ Multiplication Operator (`*`)

The **Multiplication (`*`) operator** is an arithmetic operator in Python used to **multiply two operands**. It calculates the product of two numeric values and is commonly used in mathematical calculations, finance, engineering, scientific computing, and data analysis.

In addition to numeric multiplication, the `*` operator can also be used to **repeat sequences** such as strings, lists, and tuples.

#### 📖 Topics Covered

##### 1. Introduction to the Multiplication Operator

The Multiplication operator (`*`) is used to calculate the **product** of two operands.

It is one of the most frequently used arithmetic operators in Python.

The operator can also repeat sequences such as strings, lists, and tuples.

---

##### 2. Syntax

##### Syntax

```python
operand1 * operand2
```

##### Explanation

- `operand1` → First operand.
- `*` → Multiplication operator.
- `operand2` → Second operand.

The result is the product of the two operands.

---

##### 3. Multiplication with Integers

The operator multiplies integer values.

**Example**

```python
a = 15
b = 6

print(a * b)
```

##### Output

```text
90
```

---

##### 4. Multiplication with Floating-Point Numbers

The Multiplication operator also works with decimal numbers.

**Example**

```python

```

##### Output

```text
50.0
```

---

##### 5. Multiplication with Complex Numbers

Complex numbers can also be multiplied.

**Example**

```python
a = 2 + 3j
b = 4 + 5j

print(a * b)
```

##### Output

```text
(-7+22j)
```

---

##### 6. Sequence Repetition

The `*` operator can repeat sequences.

##### String Repetition

```python
text = "Hi "

print(text * 3)
```

##### Output

```text
Hi Hi Hi
```

##### List Repetition

```python
numbers = [1, 2]

print(numbers * 3)
```

##### Output

```text
[1, 2, 1, 2, 1, 2]
```

##### Tuple Repetition

```python
colors = ("Red", "Blue")

print(colors * 2)
```

##### Output

```text
('Red', 'Blue', 'Red', 'Blue')
```

Avoid these common mistakes while using the Multiplication operator.

````python
# Multiplication
a = 10
b = 5

result = a * b
print(result)
````

````python
# Multiple Multiplication
a = 2
b = 3
c = 4

result = a * b * c
print(result)
````

````python
# Integer Multiplication
a = 100
b = 5

print(a * b)
````

````python
# Floating-Point Multiplication
a = 10.5
b = 2.5

print(a * b)
````

````python
# Integer * Float
a = 10
b = 2.5

result = a * b
print(result)
````

````python
# Negative Numbers
a = -10
b = -5

result = a * b
print(result)
````

````python
# Positive * Negative
a = 10
b = -5

result = a * b
print(result)
````

````python
# Variables
price = 100
quantity = 5

total = price * quantity
print(total)
````

````python
# User Input
a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

result = a * b
print(result)
````

````python
# Decimal Input
a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

result = a * b
print(result)
````

````python
# Multiplication Assignment
number = 10
number *= 5

print(number)
````

````python
# Multiplication in a Loop
result = 1

for number in range(1, 6):
    result = result * number
print(result)
````

````python
# Boolean Multiplication
a = True
b = 5

print(a * b)
````

````python
# Complex Number Multiplication
a = 2 + 3j
b = 4 + 5j

result = a * b
print(result)
````

````python
# Parentheses
result = (10 + 5) * 2

print(result)
````

````python
# Operator Precedence
result = 10 + 5 * 2

print(result)
````

````python
# String Multiplication
text = "Python"
result = text * 3

print(result)
````

````python
# String Repetition
text = "Python "
result = text * 3

print(result)
````

````python
# List Multiplication
numbers = [1, 2, 3]
result = numbers * 3

print(result)
````

````python
# Tuple Multiplication
numbers = (1, 2, 3)
result = numbers * 2

print(result)
````

````python
# Creating a Pattern
print("*" * 50)
````

````python
print("-" * 50)
````

````python
# Area of Rectangle
length = 10
width = 5

area = length * width
print(area)
````

````python
# Total Cost
price = 250
quantity = 4

total_cost = price * quantity
print(total_cost)
````

````python
# Distance
speed = 60
time = 3

distance = speed * time
print(distance)
````
