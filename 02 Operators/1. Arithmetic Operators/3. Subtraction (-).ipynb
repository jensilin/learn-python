#### ➖ Subtraction Operator (`-`)

The **Subtraction (`-`) operator** is an arithmetic operator in Python used to **subtract one numeric value from another**. It calculates the difference between two operands and is commonly used in mathematical calculations, finance, statistics, scientific computing, and everyday programming.

The subtraction operator works with **integers (`int`)**, **floating-point numbers (`float`)**, and **complex numbers (`complex`)**.

#### 📖 Topics Covered

##### 1. Introduction to the Subtraction Operator

The Subtraction operator (`-`) is used to find the **difference** between two numbers.

It subtracts the second operand from the first operand.

**Example**

```python
x = 50
y = 15

print(x - y)
```

##### Output

```text
35
```

---

##### 2. Syntax

##### Syntax

```python
operand1 - operand2
```

##### Explanation

- `operand1` → The value from which another value is subtracted.
- `-` → Subtraction operator.
- `operand2` → The value to subtract.

The result is the difference between the two operands.

---

##### 3. Subtraction with Integers

The Subtraction operator works with integer values.

**Example**

```python
a = 100
b = 35

print(a - b)
```

##### Output

```text
65
```

---

##### 4. Subtraction with Floating-Point Numbers

The operator also works with decimal numbers.

**Example**

```python
a = 25.8
b = 10.4

print(a - b)
```

##### Output

```text
15.4
```

---

##### 5. Subtraction with Complex Numbers

Complex numbers can also be subtracted.

**Example**

```python
a = 8 + 5j
b = 3 + 2j

print(a - b)
```

##### Output

```text
(5+3j)
```

---

##### 6. Unary Minus Operator

The minus (`-`) symbol can also be used as a **unary operator** to change the sign of a number.

**Example**

```python
number = 15

print(-number)
```

##### Output

```text
-15
```

##### 9. Common Mistakes

Avoid these common mistakes while using the Subtraction operator.

##### Using Strings Instead of Numbers

**Incorrect**

```python
print("20" - "10")
```

##### Output

```text
TypeError
```

##### Correct

```python
print(int("20") - int("10"))
```

##### Output

```text
10
```

> **Note:** The Subtraction (`-`) operator is one of the core arithmetic operators in Python. Understanding how it works with different numeric data types is essential for solving mathematical problems, performing calculations, and developing real-world applications.

````python
# Subtraction

a = 10
b = 5

result = a - b
print(result)
````

````python
# Multiple Subtraction

a = 100
b = 20
c = 10

result = a - b - c
print(result)
````

````python
# Integer Subtraction

a = 100
b = 40
print(a - b)
````

````python
# Floating-Point Subtraction

a = 10.5
b = 2.5
print(a - b)
````

````python
# Integer - Float

a = 10
b = 2.5

result = a - b
print(result)
````

````python
# Negative Numbers

a = -10
b = -5

result = a - b
print(result)
````

````python
# Positive - Negative

a = 10
b = -5

result = a - b
print(result)
````

````python
# Negative - Positive

a = -10
b = 5

result = a - b
print(result)
````

````python
# Variables

price = 1000
discount = 200

final_price = price - discount
print(final_price)
````

````python
# User Input

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))

result = a - b
print(result)
````

````python
# Decimal Input

a = float(input("Enter first number: "))
b = float(input("Enter second number: "))

result = a - b
print(result)
````

````python
# Subtraction Assignment

number = 10
number -= 3

print(number)
````

````python
# Subtraction in a Loop

number = 20

for value in range(1, 6):
    number = number - value
print(number)
````

````python
# Boolean Subtraction
#True  = 1
#False = 0

a = True
b = False
print(a - b)
````

````python
# Complex Number Subtraction

a = 5 + 7j
b = 2 + 3j

result = a - b
print(result)
````

````python
# Parentheses

result = 100 - (20 + 10)
print(result)
````

````python
result = 100 - 20 + 10
print(result)
````

````python
# Operator Precedence

result = 100 - 10 * 2
print(result)
````

````python
# Using Parentheses

result = (100 - 10) * 2
print(result)
````

````python
# Difference

age1 = 25
age2 = 21

difference = age1 - age2
print(difference)
````

````python
# Remaining Balance

balance = 5000
expense = 1250

remaining = balance - expense
print(remaining)
````

````python
# Marks Difference

highest_mark = 95
student_mark = 82

difference = highest_mark - student_mark
print(difference)
````
