#### 💾 Bytes (`bytes`) Data Type

The **`bytes`** data type is one of Python's built-in **binary data types** used to store a sequence of **bytes** (integers ranging from **0 to 255**).

A `bytes` object is **immutable**, which means once it is created, its contents cannot be modified. It is commonly used when working with binary files, network communication, image processing, encryption, and other low-level programming tasks.

Unlike strings, which store Unicode characters, `bytes` stores raw binary data.

> **Note:** This notebook focuses on the **`bytes`** data type. By the end of this notebook, you will understand how to create, access, manipulate, and use bytes objects efficiently when working with binary data in Python.

````python
# Creating Bytes

data = bytes([65, 66, 67])
print(data)
````

````python
# String to Bytes

text = "Python"
data = text.encode("utf-8")
print(data)
````

````python
# String to Bytes

text = "Python"
data = bytes(text, "utf-8")
print(data)
````

````python
# Empty Bytes

data = bytes()
print(data)
print(type(data))
````

````python
# Bytes with Numbers

data = bytes([10, 20, 30, 40, 50])
print(data)
````

````python
# Accessing Bytes

data = bytes([65, 66, 67])
print(data[0])
print(data[1])
print(data[2])
````

````python
# Negative Indexing

data = bytes([65, 66, 67, 68])
print(data[-1])
print(data[-2])
````

````python
# Bytes Slicing

data = bytes([65, 66, 67, 68, 69])
print(data[1:4])
````

````python
# Bytes are Immutable

data = bytes([65, 66, 67])
# data[0] = 90
print(data)
````

````python
# Length

data = bytes([65, 66, 67, 68])
print(len(data))
````

````python
# Checking Membership

data = bytes([65, 66, 67])
print(65 in data)
print(90 in data)
````

````python
# Bytes with for loop

data = bytes([65, 66, 67])
for value in data:
    print(value)
````

````python
# Bytes to String

data = b"Python"
text = data.decode("utf-8")
print(text)
````

````python
# ASCII Characters

data = b"ABC"
print(data)
print(data[0])
print(data[1])
print(data[2])
````

````python
# Escape Sequences

data = b"Hello\nPython"
print(data)
````

````python
# Bytes vs Bytearray

data1 = bytes([65, 66, 67])
data2 = bytearray([65, 66, 67])
print(data1)
print(data2)
````

````python
data1 = bytes([65, 66, 67])
# data1[0] = 90
````

````python
data2 = bytearray([65, 66, 67])
data2[0] = 90
print(data2)
````

````python
# Tuple to Bytes

numbers = (65, 66, 67)
data = bytes(numbers)
print(data)
````

````python
# List to Bytes

numbers = [65, 66, 67]
data = bytes(numbers)
print(data)
````
