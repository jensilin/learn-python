#### 💾 Memoryview (`memoryview`) Data Type

The **`memoryview`** data type is one of Python's built-in **binary data types** that provides a **memory view** of another binary object without copying its data.

Instead of creating a new copy of the data, a `memoryview` object allows direct access to the memory of objects such as `bytes`, `bytearray`, and other objects that support the **buffer protocol**. This makes it highly efficient when working with large binary data.

The `memoryview` data type is commonly used in file processing, networking, image processing, scientific computing, and applications where performance and memory efficiency are important.

##### 2. Characteristics of `memoryview`

Explore the important characteristics of the `memoryview` data type.

##### Characteristics

- Provides a view of existing binary data.
- Does not create a copy of the data.
- Memory efficient.
- Supports indexing and slicing.
- Useful for large binary objects.
- Changes made through the view affect the original object (if the underlying object is mutable).

**Example**

```python

view = memoryview(data)
```

##### 10. Limitations of `memoryview`

Understand the limitations of the `memoryview` data type.

##### Limitations

- Works only with objects that support the buffer protocol.
- Cannot be created directly from ordinary Python objects like lists or dictionaries.
- Mutability depends on the underlying object.
- Can be more difficult to understand than `bytes` or `bytearray` for beginners.

> **Note:** This notebook focuses on the **`memoryview`** data type. By the end of this notebook, you will understand how to efficiently access and manipulate binary data without creating unnecessary copies, making `memoryview` an excellent choice for high-performance and memory-efficient Python applications.

````python
# Creating a Memoryview

data = bytearray([65, 66, 67, 68])
view = memoryview(data)
print(view)
````

````python
# Checking Type

data = bytearray([65, 66, 67])
view = memoryview(data)
print(type(view))
````

````python
# Accessing Elements

data = bytearray([65, 66, 67, 68])
view = memoryview(data)
print(view[0])
print(view[1])
print(view[2])
````

````python
# Negative Indexing

data = bytearray([65, 66, 67, 68])
view = memoryview(data)
print(view[-1])
print(view[-2])
````

````python
# Slicing

data = bytearray([65, 66, 67, 68, 69])
view = memoryview(data)
result = view[1:4]
print(result)
````

````python
# Slicing

data = bytearray([65, 66, 67, 68, 69])
view = memoryview(data)
result = view[1:4]
print(result)
````

````python
# Memoryview to Bytes

data = bytearray([65, 66, 67])
view = memoryview(data)
result = bytes(view)
print(result)
````

````python
# Memoryview to Bytearray

data = bytearray([65, 66, 67])
view = memoryview(data)
result = bytearray(view)
print(result)
````

````python
# Memoryview with Bytearray

data = bytearray([65, 66, 67])
view = memoryview(data)
print(view[0])
print(view[1])
print(view[2])
````

````python
# Modifying Data

data = bytearray([65, 66, 67])
view = memoryview(data)
view[0] = 90
print(data)
````

````python
# Length

data = bytearray([65, 66, 67, 68])
view = memoryview(data)
print(len(view))
````

````python
# Memoryview with for loop

data = bytearray([65, 66, 67])
view = memoryview(data)
for value in view:
    print(value)
````

````python
# String to Memoryview

text = "Python"
data = text.encode("utf-8")
view = memoryview(data)
print(view)
print(view[0])
````

````python
# Checking Membership

data = bytearray([65, 66, 67])
view = memoryview(data)
print(65 in view)
print(90 in view)
````

````python
# Memoryview does not copy the data

data = bytearray([65, 66, 67])
view = memoryview(data)
data[0] = 90
print(view[0])
````

````python
# Bytes vs Bytearray vs Memoryview

data1 = bytes([65, 66, 67])
data2 = bytearray([65, 66, 67])
data3 = memoryview(data2)
print(data1)
print(data2)
print(data3)
````
