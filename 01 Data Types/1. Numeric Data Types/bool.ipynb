#### ✅ Boolean (`bool`) Data Type

The **Boolean (`bool`)** data type is one of Python's built-in numeric data types. It is used to represent **logical values** and has only **two possible values**: `True` and `False`.

The `bool` data type is primarily used in **decision-making**, **conditional statements**, **loops**, and **logical operations**. Internally, Python treats `True` as `1` and `False` as `0`, making the Boolean data type a subclass of the `int` data type.

Boolean values are immutable and are commonly produced as the result of comparison and logical operations.

````python
#Creating Boolean Values

is_student = True
is_working = False

print(is_student)
print(is_working)
````

````python
#Checking Data Type

Jensilin = True
print(type(Jensilin))
````

````python
#Boolean from Comparison

print(10 > 5)
print("Jensilin" == "jensilin")
print(10!=9)
print(80<=75)
````

````python
#Using bool()
print(bool(1))
print(bool(0))
print(bool("Jensilin"))
print(bool(""))
````

````python
#Truthy Values
print(bool(100))
print(bool(-5))
print(bool([1,2,3,4]))
print(bool({"Jensilin":75}))
````

````python
#Falsy Values
print(bool(None))
print(bool(""))
print(bool([]))
print(bool({}))
print(bool(0))
````

````python
#Boolean Operators
print(True and False)
print(True or False)
print(not True)
print(not False)
````

````python
#Comparison Operators

a = 9
b = 15

print(a>b)
print(a<b)
print(a>=b)
print(a<=b)
print(a==b)
print(a!=b)
````

````python
#Using Boolean in if
is_logged_in = True

if (is_logged_in): print("Welcome!")
else : print("Please login.")
````

````python
#Boolean with Variables

age = 18

can_vote = age>=18

print(can_vote)
````

````python
#Boolean from User Input

password = "2026$Password"

is_valid = len(password)>=8

print(is_valid)
````

````python
#bool is a Subclass of int
print(True + True)
print(False - True)
print(int(True))
print(int(False))
````

````python
#Equality vs Identity
print(True == True)
print(False == False)

print(True is True)
print(False is False)
print(False is not False)
````

````python
#Boolean in Collections

number = []

if number: print("List is not empty.")
else : print("List is empty")
````

````python
#Real-world Example

balance = 10000
withdraw = 5000

has_balance = withdraw<=balance
print(has_balance)
````
