#### 🔤 String (`str`) Data Type

The **String (`str`)** data type is one of Python's built-in **sequence data types**. It is used to represent **textual data** and consists of a sequence of **Unicode characters** enclosed within single quotes (`' '`), double quotes (`" "`), or triple quotes (`''' '''` or `""" """`).

Strings are **immutable**, which means their contents cannot be modified after creation. Python provides a rich collection of built-in functions and methods for performing various string operations such as searching, formatting, splitting, joining, replacing, and case conversion.

Strings are one of the most frequently used data types in Python and play an important role in text processing, data manipulation, file handling, web development, and many other real-world applications.

````python
#String Indexing
name = "Jensilin"

print(name[0])
print(name[1])
print(name[2])
print(name[5])
````

````python
#Negative Indexing
name = "Jensilin"

print(name[-1])
print(name[-2])
print(name[-3])
print(name[-7])
````

````python
#String Slicing
name = "Jensilin"

print(name[1:5])
````

````python
#Slicing from Beginning
name = "Jensilin"

print(name[:5])
````

````python
#Slicing to End
name = "Jensilin"

print(name[2:])
````

````python
#Complete String
name = "Jensilin"

print(name[:])
````

````python
#Step Slicing
name = "Jensilin"

print(name[::2])
````

````python
#Reverse String Using Slicing
name = "Jensilin"

print(name[::-1])
````

````python
#String Concatenation

fn = "Jemsilin"
ln = "Jino"

fl = fn+" "+ln

print(fl)
````

````python
#String Repetition
name = "Jensilin"

print(name*4)
````

````python
#Membership Operator 
#in Operator
name = "Jensilin"

print('J' in name)
print('z' in name)
````

````python
#Membership Operator 
#not in Operator

name = "Jensilin"

print('J' not in name)
print('z' not in name)
````

````python
#Comparing Strings
a = "Python"
b = "Python"

print(a == b)
````

````python
#String Comparison
a = "Apple"
b = "Banana"

print(a == b)
print(a != b)
print(a < b)
print(a > b)
````

````python
#Iterating Through a String
name = "Jensilin"

for i in name:
    print(i)
````

````python
#String with Numbers
age = 23

message = "My age is " + str(age)

print(message)
````

````python
#String Formatting with f-String
name = "Jensilin"
age = 23

message = f"My name is {name} and I am {age} years old."

print(message)
````

````python
#Escape Characters
message = "Python is \"easy\" to learn."

print(message)
````

````python
#New Line
message = "Hello\nPython"

print(message)
````
