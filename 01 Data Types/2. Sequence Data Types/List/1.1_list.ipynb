#### 📋 List (`list`) Data Type

The **List (`list`)** is one of Python's built-in **sequence data types**. It is used to store an **ordered collection of multiple values** in a single variable. A list can contain elements of the **same or different data types**, making it one of the most flexible and widely used data structures in Python.

Lists are **mutable**, which means their elements can be added, removed, or modified after the list has been created. They also allow duplicate values and support indexing, slicing, iteration, and various built-in methods.

Lists are commonly used for storing collections of data, processing information, managing records, and implementing various data structures and algorithms.

````python
#Creating a List

n = [10, "Jensilin", 23]
print(n)
````

````python
#Different Types of Lists

numbers = [1, 2, 3]

names = ["Jensilin", "Python", "Java"]

mixed = [10, "Python", 3.14, True]

print(numbers)
print(names)
print(mixed)
````

````python
#Checking the Data Type

fruits = ["Mango", "Banana", "Orange"]

print(type(fruits))
````

````python
#List Length
names = ["Jensilin", "Python", "Java"]
print(len(names))
````

````python
#Accessing Elements (Positive Indexing)
names = ["Jensilin", "Python", "Java"]

print(names[0])
print(names[1])
print(names[2])
````

````python
#Negative Indexing
names = ["Jensilin", "Python", "Java"]

print(names[-1])
print(names[-2])
print(names[-3])
````

````python
#Slicing a List

numbers = [10, 20, 30, 40, 50]

print(numbers[1:4])
print(numbers[0:3])
print(numbers[2:4])
````

````python
#Slicing from Beginning
numbers = [10, 20, 30, 40, 50]

print(numbers[:3])
````

````python
#Slicing to End
numbers = [10, 20, 30, 40, 50]

print(numbers[2:])
````

````python
#Complete Slice
numbers = [10, 20, 30, 40, 50]

print(numbers[:])
````

````python
#Step Slicing
numbers = [10, 20, 30, 40, 50, 60]

print(numbers[::2])
````

````python
#Reverse Using Slicing
numbers = [10, 20, 30, 40]

print(numbers[::-1])
````

````python
#Nested Lists
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

print(matrix)
````

````python
#Accessing Nested List Elements
matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

print(matrix[0][1])
print(matrix[2][0])
print(matrix[1][1])
````

````python
#Updating an Element
names = ["Jensilin", "Python", "Java"]
names[2] = "Data Mining"
print(names)
````

````python
#List Concatenation

l1 = [10, 20, 30]
l2 = [60, 70, 80, 90]

r = l1 + l2
print(r)
````

````python
#List Repetition

me = ["Jensilin",23]
print(me*5)
````

````python
#Membership Operator
names = ["Jensilin", "Python", "Data Mining"]

print("Jensilin" in names)
print("Java" in names)
````

````python
#Iterating Through a List
names = ["Jensilin", "Python", "Data Mining"]

for i in names:
    print(i)
````

````python
#Iterating Using Index
names = ["Jensilin", "Python", "Data Mining"]

for i in range(len(names)):
    print(names[i])
````

````python
#Comparing Lists
list1 = [1, 2, 3]
list2 = [1, 2, 3]

print(list1 == list2)
````

````python
#Empty List
numbers = []

print(numbers)
print(len(numbers))
````

````python
#List Packing
numbers = [10, 20, 30]

print(numbers)
````

````python
#List Unpacking
numbers = [10, 20, 30]

a, b, c = numbers

print(a)
print(b)
print(c)
````

````python
#Real-world Example
students = ["Alice", "Bob", "Charlie", "David"]

print("Total Students:", len(students))
print("First Student:", students[0])
print("Last Student:", students[-1])
````
