#### ❄️ Frozen Set (`frozenset`) Data Type

The **Frozen Set (`frozenset`)** is one of Python's built-in **set data types**. It is an **immutable version of a set**, meaning that once a `frozenset` is created, its elements **cannot be added, removed, or modified**.

Like a normal set, a `frozenset` stores **unique elements**, does **not allow duplicate values**, and supports mathematical set operations such as union, intersection, difference, and symmetric difference.

Because `frozenset` is immutable, it is **hashable**, which means it can be used as a **dictionary key** or as an element of another set.

````python
#Creating a Frozenset

fruits = frozenset({"Apple", "Banana", "Mango"})
print(fruits)
````

````python
#Frozenset with Numbers

numbers = frozenset({10, 20, 30, 40, 50})
print(numbers)
````

````python
#Different Data Types

data = frozenset({"Jensilin", 21, 50.5, True})
print(data)
````

````python
#List to Frozenset

numbers = [10, 20, 30, 20, 10]
numbers_set = frozenset(numbers)
print(numbers_set)
````

````python
#Tuple to Frozenset

numbers = (10, 20, 30, 20, 10)
numbers_set = frozenset(numbers)
print(numbers_set)
````

````python
#Empty Frozenset

items = frozenset()
print(items)
print(type(items))
````

````python
#Duplicate Values

numbers = frozenset({10, 20, 10, 30, 20, 40})
print(numbers)
````

````python
#Length of Frozenset

fruits = frozenset({"Apple", "Banana", "Mango", "Orange"})
print(len(fruits))
````

````python
#Checking Membership

fruits = frozenset({"Apple", "Banana", "Mango"})
print("Apple" in fruits)
print("Orange" in fruits)
````

````python
#Frozenset with for loop

fruits = frozenset({"Apple", "Banana", "Mango"})
for fruit in fruits:
    print(fruit)
````

````python
#Frozenset does not support indexing

fruits = frozenset({"Apple", "Banana", "Mango"})
# print(fruits[0])
````

````python
#Frozenset is Immutable

fruits = frozenset({"Apple", "Banana", "Mango"})
# fruits.add("Orange")
print(fruits)
````

````python
#Set vs Frozenset

normal_set = {"Apple", "Banana", "Mango"}
fixed_set = frozenset({"Apple", "Banana", "Mango"})
print(normal_set)
print(fixed_set)
````

````python
#Checking Type

fruits = frozenset({"Apple", "Banana", "Mango"})
print(type(fruits))
````

````python
#String to Frozenset

letters = frozenset("Python")
print(letters)
````
