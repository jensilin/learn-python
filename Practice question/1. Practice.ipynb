##### Question 1: Student Grade Management System

Write a Python program to manage student records using a dictionary.

Requirements:
- Store student names as keys and marks as values.
- Allow the user to:
  - Add a student.
  - Update marks.
  - Delete a student.
  - Search for a student.
  - Display all students sorted by marks (highest to lowest).
- Display the average, highest, and lowest marks.

**Example Input**
```text
Enter number of students: 3
Enter student name: Alice
Enter marks: 85
Enter student name: Bob
Enter marks: 92
Enter student name: Charlie
Enter marks: 78
Search student: Bob
```

**Expected Output**
```text
Student Found: Bob
Marks: 92

Students (Highest to Lowest)
Bob : 92
Alice : 85
Charlie : 78

Average Marks : 85.00
Highest Marks : 92
Lowest Marks : 78
```

````python
students = {}

def add_student():
    n = int(input("Enter number of students: "))

    for i in range(n):
        student_name = input("Enter student name: ").lower()
        student_marks = int((input("Enter marks: ")))
        students[student_name] = student_marks

    print(f'Total studends : {students}')

def search_student():
    s_n = input("Search student: ").lower()
    if s_n in students.keys():
        print(students[s_n])
    else: print(f'{s_n} not found in student records.')

def delete_student():
    d_n = input("Student to be deleted: ").lower()
    for i in students.keys():
        if i == d_n:
            students.pop(i)
            break
    print(f'The student {d_n} is deleted.')
    print(f'Total studends : {students}.')

def sort_max_min_percentage():
    st = sorted(students)
    print("Students Highest to Lowest.")
    for i in st:
        print(f'{i} : {students[i]}')
    print("")
    print(f'Average Marks : {(sum(students.values())/len(students))}')
    print(f'Highest Marks : {max(students.values())}')
    print(f'Lowest Marks : {min(students.values())}')
````

````python
add_student()
````

````python
search_student()
````

````python
delete_student()
````

````python
sort_max_min_percentage()
````

Create a payroll management system.

Requirements:
- Store employee details (Name, Basic Salary, Experience).
- Calculate:

  - DA = 15% of Basic Salary
  - Bonus:
    - Experience ≥ 5 years → 10%
    - Otherwise → 5%
- Calculate Gross Salary.
- Display payroll in a formatted table.

**Example Input**
```text
Employee Name: John
Basic Salary: 50000
Experience: 6
```

**Expected Output**
```text
Employee Payroll

Name          : John
Basic Salary  : 50000
HRA           : 10000
DA            : 7500
Bonus         : 5000
Gross Salary  : 72500
```

````python
name = "John"
salary = 50000
experience = 6
def payroll():
    print("Employee Payroll")
    print("")
    hra = salary * 0.2
    da = salary * 0.15
    per = 0
    if experience >=5 :
        per = salary * 0.10
    else:
        per = salary * 0.05

    print(f'Name          : {name}')
    print(f'Basic Salary  : {salary}')
    print(f'HRA           : {int(hra)}')
    print(f'DA            : {int(da)}')
    print(f'Bonus         : {int(per)}')
    print(f'Gross Salary  : {int(salary+hra+da+per)}')
````

##### Question 3: Library Book Management System

Develop a Library Management System.

Requirements:
- Add books.
- Issue books.
- Return books.
- Search books by title.
- Display available books.
- Prevent issuing unavailable books.
- Store all book details in a dictionary.

**Example Input**
```text
Add Book: Python Programming
Issue Book: Python Programming
Search Book: Python Programming
```

**Expected Output**
```text
Book Added Successfully.

Book Issued Successfully.

Book Details
Title : Python Programming
Status : Issued
```

````python
books = {}

def add_books():
    inp = input("Add Book: ").lower()
    books[inp] = "not issued"
    print(f'{inp.title()} Added.')

def issue_books():
    inp = input("Issue Book: ").lower()
    for i in books.keys():
        if i == inp:
            books[i]="issued"
            print(f'{i.title()} Issued Successfully.')
            break
    else:
        print("Book not found.")

def return_books():
    inp = input("Return Book: ").lower()
    for i in books.keys():
        if i == inp:
            books[i]="not issued"
            print(f'{i.title()} returned successfully.')
            break
    else:
        print("Book not from this library.")

def display_books():
    print("Book Details")
    for i in books.keys():
        print(f'Title : {i.title()}')
        print(f'Status : {books[i].title()}')
````

````python
add_books()
````

````python
issue_books()
````

````python
return_books()
````

````python
display_books()
````

##### Question 4: Password Strength Analyzer

Write a program to validate password strength.

Requirements:
A password is considered strong only if it contains:
- Minimum 10 characters
- At least one uppercase letter
- At least one lowercase letter
- At least one digit
- At least one special character
- No spaces
- Should not contain the username

Display all validation failures.

**Example Input**
```text
Username: john
Password: John123
```

**Expected Output**
```text
Password is Weak

Reasons:
- Password must contain at least 10 characters.
- Password must contain at least one special character.
```

````python
password = input("Enter the password: ")
````

````python
if len(password)>=10:
    up = 0
    lw = 0
    num = 0
    spec = 0
    for i in password:
        if i.isupper():
            if up == 0:
                up+=1
            else : continue
        elif i.islower():
            if lw == 0:
                lw+=1
            else : continue
        elif '0'<=i<='9':
            if num == 0:
                num+=1
            else : continue
        else:
            if spec == 0:
                spec+=1
            else : continue

    if (up+lw+num+spec) == 4:
        print("Password created.")
    else :
        if up!=1 :
            print("Password must contain at least one upper case character.")
        elif lw!=1:
            print("Password must contain at least one lower case character.")
        elif num!=1:
            print("Password must contain at least one digits.")
        elif spec!=1:
            print("Password must contain at least one special character.")
else :
    print("Password must contain at least 10 characters.")
````
